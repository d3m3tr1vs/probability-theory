## Задача 1.8
В конкурсе по 5 номинациям участвуют 10 кино-фильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

а) Порядок важен, т.к. разные призы; с повторениями. Используем размещения с повторениями: $\overline{A_{10}^{5}}=10^5=100000$

```python
10**5
```

б) Порядок не важен, т.к. призы одинаковые. Используем сочетания с повторениями: $\overline{C^5_{10}}=C^5_{10+5-1}=C^5_{14}=2002$

```python
import scipy.special as sc
sc.comb(14,5)
```

## Задача 1.9
Сколько существует семизначных чисел, состоящих из цифр 4, 5 и 6, в которых цифра 4 повторяется 3 раза, а цифры 5 и 6 - по 2 раза?

Используем перестановки с повторениями: $P_7(3;2;2)=\frac{7!}{3!2!2!}=210$

```python
sc.factorial(7)/(sc.factorial(3)*sc.factorial(2)*sc.factorial(2))
```

## Задача 1.10
Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

а) $P(A)=\frac{m}{n}=\frac{1}{A_6^3}=\frac{1}{6\cdot5\cdot4}=\frac{1}{120}$

```python
1/sc.perm(6,3)
```

б) $P(B)=\frac{1}{P_6}=\frac{1}{6!}=\frac{1}{720}$

```python
1/sc.factorial(6)
```

## Задача 1.11
Используя условие примера 1.10, найти вероятность того, что получится слово «АНАНАС», если на отдельных карточках написаны три буквы А, две буквы Н и одна буква С.

$P(B)=\frac{m}{n}=\frac{P_3\cdot P_2}{P_6}=\frac{3!2!}{6!}=\frac{1}{60}$

```python
sc.factorial(3) * sc.factorial(2) / sc.factorial(6)
```

## Задача 1.16
За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

а) $P(A)=\frac{m_1}{n}=\frac{2(5!)^2}{10!}=\frac{120^2\cdot2}{3628800}=0,00794$

```python
2*sc.factorial(5)**2/sc.factorial(10)
```

б) $P(B)=\frac{m_2}{n}=\frac{5!2^5}{10!}=\frac{120\cdot32}{3628800}=0,00106$

```python
2**5*sc.factorial(5)/sc.factorial(10)
```
