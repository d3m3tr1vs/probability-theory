### Задача 1.8
>В конкурсе по 5 номинациям участвуют 10 кино-фильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

а) Порядок важен, т.к. разные призы; с повторениями. Используем размещения с повторениями: $\overline{A_{10}^{5}}=10^5=100000$

б) Порядок не важен, т.к. призы одинаковые. Используем сочетания с повторениями: $\overline{C^5_{10}}=C^5_{10+5-1}=C^5_{14}=2002$

```python
import scipy.special as sc
round(sc.comb(14,5),4)
```

### Задача 1.10
>Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

а) Событие A - получилось слово "ТОР"

$P(A)=\frac{m}{n}=\frac{1}{A_6^3}=\frac{1}{6\cdot5\cdot4}=\frac{1}{120}$

```python
round(1/sc.perm(6,3),4)
```

б) Событие B - "ТЕОРИЯ"

$P(B)=\frac{1}{P_6}=\frac{1}{6!}=\frac{1}{720}$

```python
round(1/sc.factorial(6),4)
```

### Задача 1.16
>За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

```python
round(2**5*sc.factorial(5)/sc.factorial(10),5)
```

### Задача 1.48. 
>1.48. В старинной игре в кости необходимо было для выигрыша
получить при бросании трех игральных костей сумму очков, превосходящую 10. Найти вероятности: а) выпадения 11 очков; б) выигрыша.

а) Рассмотрим событие A - в сумме выпало 11 очков

б) Событие B - в сумме выпало больше 10, т.е. [11, 18]

```python
import itertools
outcomes = list(itertools.product(range (1,7), repeat=3))
len_outcomes = len(outcomes)
#print (outcomes)
A = sum(sum(x)==11 for x in outcomes)
B = sum(sum(x)>10 for x in outcomes)
print(f"P(A)={A/len_outcomes}\nP(B)={B/len_outcomes}")
```

# Практика 4. Условная вероятность
### Задача.
>В одной урне 15 белых и 5 черных шаров, а в другой - 6 белых и 1 черный шар. Из первой урны случайным образом вынимают 3 шара и опускают во вторую урну. После этого из второй урны также случайно вынимают 1 шар. Найти вероятность того, что шар, вынутый из второй урны - белый.
>Решить классически и смоделировать решение задачи.
#### Классически.
Рассмотрим событие A - из второй урны вынут белый шар.

$H_k$ - гипотеза "из первой во вторую урну перенесено ровно $k$ белых шаров", $k=0,1,2,3$.

$P(H_k)=\frac{C^k_{15}C^{3-k}_5}{C^3_{20}}$

$P(A|H_k)=\frac{6+k}{10}$

$P(A)=\sum\limits^3_{k=0}P(H_k)P(A|H_k)=\sum\limits^3_{k=0}\frac{C^k_{15}C^{3-k}_5}{C^3_{20}}\cdot\frac{6+k}{10}$

```python
PA=0
for k in range(4):
    PA+=sc.comb(15,k)*sc.comb(5,3-k)*(6+k)/(sc.comb(20,3)*10)
PA
```

#### Моделированием.

```python
import random
count=0
N=10**7
urn1=['w'] * 15 + ['b'] * 5
urn2=['w'] * 6 + ['b'] * 1
for i in range(N):
    transferred=random.sample(urn1,3)
    if random.choice(urn2+transferred)=='w':
        count+=1
count/N
```
